```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

trades = pd.read_csv("data/historical_data.csv")
sentiment = pd.read_csv("data/fear_greed_index.csv")

trades.columns = trades.columns.str.strip()
sentiment.columns = sentiment.columns.str.strip()
```

```python
trades['Timestamp IST'] = pd.to_datetime(
    trades['Timestamp IST'],
    format='%d-%m-%Y %H:%M'
)

sentiment['date'] = pd.to_datetime(
    sentiment['date']
)
```

```python
trades['Date'] = trades['Timestamp IST'].dt.date

sentiment['Date'] = sentiment['date'].dt.date
```

```python
merged = trades.merge(
    sentiment[['Date','classification']],
    on='Date',
    how='left'
)

merged.rename(
    columns={
        'classification':'Classification'
    },
    inplace=True
)

print("Merged Shape:", merged.shape)
merged.head()
```

```python
merged['is_profit'] = merged['Closed PnL'] > 0

merged['abs_pnl'] = merged['Closed PnL'].abs()

merged['trade_volume'] = merged['Size USD']
```

```python
print(merged.info())

print("\nMissing Values:")
print(merged.isnull().sum())

print("\nSummary Statistics:")
print(merged.describe())
```

```python
print(
    merged['Classification']
    .value_counts()
)
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    x='Classification',
    data=merged
)

plt.title('Sentiment Distribution')

plt.show()
```

```python
avg_pnl = merged.groupby(
    'Classification'
)['Closed PnL'].mean()

print(avg_pnl)
```

```python
plt.figure(figsize=(8,5))

avg_pnl.plot(kind='bar')

plt.title('Average PnL by Sentiment')

plt.ylabel('Average Closed PnL')

plt.show()
```

```python
median_pnl = merged.groupby(
    'Classification'
)['Closed PnL'].median()

print(median_pnl)
```

```python
win_rate = merged.groupby(
    'Classification'
)['is_profit'].mean()*100

print(win_rate)
```

```python
plt.figure(figsize=(8,5))

win_rate.plot(kind='bar')

plt.title('Win Rate (%) by Sentiment')

plt.ylabel('Win Rate %')

plt.show()
```

```python
trade_count = merged.groupby(
    'Classification'
).size()

print(trade_count)
```

```python
plt.figure(figsize=(8,5))

trade_count.plot(kind='bar')

plt.title('Trade Count by Sentiment')

plt.show()
```

```python
side_analysis = pd.crosstab(
    merged['Classification'],
    merged['Side']
)

print(side_analysis)
```

```python
side_analysis.plot(
    kind='bar',
    figsize=(10,5)
)

plt.title(
    'Long vs Short by Sentiment'
)

plt.show()
```

```python
top_traders = (
    merged.groupby('Account')
    ['Closed PnL']
    .sum()
    .sort_values(
        ascending=False
    )
    .head(20)
)

print(top_traders)
```

```python
plt.figure(figsize=(12,6))

top_traders.plot(kind='bar')

plt.title(
    'Top 20 Traders by Total PnL'
)

plt.ylabel('Total PnL')

plt.show()
```

```python
coin_activity = (
    merged['Coin']
    .value_counts()
    .head(10)
)

print(coin_activity)
```

```python
plt.figure(figsize=(10,5))

coin_activity.plot(kind='bar')

plt.title('Top Traded Coins')

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.boxplot(
    x='Classification',
    y='Closed PnL',
    data=merged
)

plt.title(
    'PnL Distribution by Sentiment'
)

plt.show()
```

```python
volume_analysis = merged.groupby(
    'Classification'
)['trade_volume'].mean()

print(volume_analysis)
```

```python
plt.figure(figsize=(8,5))

volume_analysis.plot(
    kind='bar'
)

plt.title(
    'Average Trade Volume by Sentiment'
)

plt.show()
```

```python
numeric_cols = merged.select_dtypes(
    include=np.number
)

plt.figure(figsize=(12,8))

sns.heatmap(
    numeric_cols.corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title(
    'Correlation Heatmap'
)

plt.show()
```

```python
merged.to_csv(
    "data/processed_data.csv",
    index=False
)

print("Saved Successfully")
```

1. Greed sentiment produced higher average trader profits than Fear.

2. Traders were more active during Greed periods.

3. Long positions dominated during Greed sentiment.

4. Fear sentiment showed lower win rates.

5. A small group of traders generated most of the total profits.

6. Trading volume increased significantly during Greed periods.
